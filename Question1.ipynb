```python
import tkinter as tk
from tkinter import messagebox, simpledialog, ttk
```

```python
class BudgetCalulation:
  def __init__(self, gross_monthly_income):
    self.gross_monthly_income = gross_monthly_income

  
  def calculate_after_tax_income(self):
    yearly_income = self.gross_monthly_income * 12
    if yearly_income <237_101:
      tax_yearly = yearly_income * 0.18
      monthly_tax = round(tax_yearly/12, 2)
    elif yearly_income > 237100 and yearly_income < 370501:
      payable_tax = yearly_income - 237100
      taxed_amount = payable_tax * 0.26
      total_tax = 42678 + taxed_amount
      tax_yearly = yearly_income - total_tax 
      monthly_tax = round(tax_yearly/12, 2)
    elif yearly_income > 370500 and yearly_income < 512801:
      payable_tax = yearly_income - 370500
      taxed_amount = payable_tax * 0.31
      total_tax = 77362 + taxed_amount
      tax_yearly = yearly_income - total_tax 
      monthly_tax = round(tax_yearly/12, 2)
    elif yearly_income > 512800 and yearly_income < 673001:
      payable_tax = yearly_income - 512800
      taxed_amount = payable_tax * 0.36
      total_tax = 121475 + taxed_amount
      tax_yearly = yearly_income - total_tax 
      monthly_tax = round(tax_yearly/12, 2)
    elif yearly_income > 673000 and yearly_income < 857900:
      payable_tax = yearly_income - 673000
      taxed_amount = payable_tax * 0.39
      total_tax = 179147 + taxed_amount
      tax_yearly = yearly_income - total_tax 
      monthly_tax = round(tax_yearly/12, 2)
    elif yearly_income > 857900 and yearly_income < 1_817000:
      payable_tax = yearly_income - 857900
      taxed_amount = payable_tax * 0.41
      total_tax = 251258 + taxed_amount
      tax_yearly = yearly_income - total_tax 
      monthly_tax = round(tax_yearly/12, 2)
    elif yearly_income > 1_817000:
      payable_tax = yearly_income - 1_817000
      taxed_amount = payable_tax * 0.45
      total_tax = 644489 + taxed_amount
      tax_yearly = yearly_income - total_tax 
      monthly_tax = round(tax_yearly/12, 2)

    return monthly_tax
  
  def calculate_expenses(self):
    net_income = self.calculate_after_tax_income()
    expenses = {"Utilities": net_income * 0.05,
                "Rent/Housing": net_income * 0.15,
                "Transportation": net_income * 0.30,
                "Healthcare": net_income * 0.03,
                "Groceries": net_income * 0.10,
                "Communication": net_income * 0.02,
    }
    return expenses
  
  def calculate_total_expenses(self):
    expenses = self.calculate_expenses()
    total = sum(expenses.values())
    return total
  
  def net_income_after_expenses(self):
    net_income_after_tax = self.calculate_after_tax_income()
    total_expenses = self.calculate_total_expenses()
    return net_income_after_tax - total_expenses
```

```python
class BudgetUI:
  def __init__(self, root):
    self.root = root
    self.root.title("Budget Portal")
    self.root.geometry("400x300")

    self.center_window()
    self.style = ttk.Style()
    self.style.configure("TButton", font=("Arial", 12), padding=10)
    frame = ttk.Frame(root, padding="20 20 20 20")
    frame.pack(fill=tk.BOTH, expand=True)

    self.button_create_entry = ttk.Button(frame, text="Create New Entry", command=self.create_new_entry,style="TButton")
    self.button_create_entry.pack(pady=20, fill=tk.X)

    self.button_exit = ttk.Button(frame, text="Exit", command=self.exit_program,style="TButton")
    self.button_exit.pack(fill=tk.X)

  def center_window(self):
    self.root.update_idletasks()
    width = self.root.winfo_width()
    height = self.root.winfo_height()
    screen_width = self.root.winfo_screenwidth()
    screen_height = self.root.winfo_screenheight()
    x = (screen_width - width) // 2
    y = (screen_height - height) // 2
    self.root.geometry(f'+{x}+{y}')


  def get_user_code(self):
    dialog = CustomInputDialog(self.root, "User Code", "Enter your user code:")
    user_code = dialog.result

    if user_code is None or not user_code.strip():
      messagebox.showerror("Invalid Input", "User code cannot be empty.")
      return self.get_user_code()
    return user_code
  
  def get_income_before_tax(self):
    dialog = CustomInputDialog(self.root, "Income", "Enter your income before tax:")
    income = dialog.result

    if income is None:
      return None
        
    try:
      income = int(income)
      if income <= 0:
        raise ValueError
      return income
    except ValueError:
      messagebox.showerror("Invalid Input", "Please enter a valid positive income before tax.")
      return self.get_income_before_tax()
  
  def create_new_entry(self):
    user_code = self.get_user_code()
    if user_code is None:
      return
    
    income_before_tax = self.get_income_before_tax()
    if income_before_tax is None:
      return
    
    budget = BudgetCalulation(income_before_tax)
    self.show_budget_summary(user_code, budget)



  def show_budget_summary(self, user_code, budget):
    def create_progress_bar(parent, value, max_value):
      progress = ttk.Progressbar(parent, length=400, mode='determinate', maximum=max_value, value=value)
      progress.pack(pady=5)
      percent_label = tk.Label(parent, text=f"{value/max_value*100:.1f}%")
      percent_label.pack()
      return progress
    
    def create_section(parent, title, bg_color):
      frame = tk.Frame(parent, bg=bg_color, padx=10, pady=5, relief="solid", borderwidth=1)
      frame.pack(fill="x", pady=3)
      tk.Label(frame, text=title, font=("Helvetica", 12, "bold")).pack(pady=2)
      return frame
    

    summary_window = tk.Toplevel(self.root)
    summary_window.title("Monthly Summary")
    summary_window.geometry("500x800")

    tk.Label(summary_window, text="Monthly Budget", font=("Helvetica", 14, "bold")).pack(pady=10)

    #Income Section#
    income_frame = create_section(summary_window, "Income","#858487")
    income_after_tax = budget.calculate_after_tax_income()
    tk.Label(income_frame, text=f"Income before tax: R{budget.gross_monthly_income:.2f}").pack(pady=2)
    tk.Label(income_frame, text=f"Income after tax: R{income_after_tax:.2f}").pack(pady=2)

    #Expenses Section#
    expenses_frame = create_section(summary_window, "Expenses", "#858487")
    expenses = budget.calculate_expenses()
    total_expenses = sum(expenses.values())

    for category, amount in expenses.items():
      cat_frame = tk.Frame(expenses_frame)
      cat_frame.pack(fill='x', pady=2)
      tk.Label(cat_frame, text=f"{category}: R{amount:.2f}", width=20, anchor='w').pack(side='left')
      create_progress_bar(cat_frame, amount, income_after_tax)

    tk.Label(expenses_frame, text=f"Total Expenses: R{total_expenses:.2f}", font=("Helvetica", 10, "bold")).pack(pady=5)

    #Net Section#
    net_frame = create_section(summary_window, "Net", "#858487")
    net_income = budget.net_income_after_expenses()
    tk.Label(net_frame, text=f"Net income after expenses: R{net_income:.2f}").pack(pady=2)
    create_progress_bar(net_frame, net_income, income_after_tax)

  def exit_program(self):
    self.root.quit()
```

```python
class CustomInputDialog(tk.Toplevel):
  def __init__(self, parent, title, prompt):
    super().__init__(parent)
    self.title(title)
    self.geometry("400x200")
    self.resizable(False, False)
    self.update_idletasks(); x = (self.winfo_screenwidth() - self.winfo_width()) // 2; y = (self.winfo_screenheight() - self.winfo_height()) // 2
    self.geometry(f'+{x}+{y}') 
    frame = ttk.Frame(self, padding="20 20 20 20")
    frame.pack(fill=tk.BOTH, expand=True) 
    ttk.Label(frame, text=prompt).pack(pady=(0, 10))
    self.entry = ttk.Entry(frame); self.entry.pack(fill=tk.X, pady=(0, 20)); self.entry.focus_set()
    ttk.Button(frame, text="Submit", command=self.submit).pack(side=tk.RIGHT, padx=(10, 0))
    ttk.Button(frame, text="Cancel", command=self.destroy).pack(side=tk.RIGHT)
    self.entry.bind('<Return>', lambda e: self.submit())
    self.transient(parent); self.grab_set(); self.wait_window(self)
    
  def submit(self):
    self.result = self.entry.get(); self.destroy()
```

```python
def main():
  root = tk.Tk()
  app = BudgetUI(root)
  root.mainloop()
```
